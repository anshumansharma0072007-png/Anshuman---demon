```python
arr = np.array([1,4,9,16])
print(np.sqrt(arr))
print(np.log(arr))
print(np.square(arr))
print(np.exp(arr))
```

BINARY UNIVERSAL FUNCTION

```python
a = np.array([1,2,3,4])
b = np.array([1,2,3,4])
print(np.add(a,b))
print(np.subtract(a,b))
print(np.multiply(a,b))
print(np.divide(a,b))
print(np.maximum(a,b))
print(np.minimum(a,b))
```

ARRAY OPRATION

```python
a = np.array([1,2,3])
b = np.array([4,5,6])
print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a%b)
print(a**b)
```

```python
marks = np.array([45,70,85,30,90])
passed = marks[marks >= 40]
print(passed)
```

AGGRIGATION OPERATION

```python
arr = np.array([10,20,30,40,50])
print(np.sum(arr))
print(np.min(arr))
print(np.max(arr))
print(np.mean(arr))
print(np.median(arr))
print(np.std(arr))
print(np.var(arr))
```

```python
arr = np.array([[1,2,3],[4,5,6]])
print(np.sum(arr,axis=0))
print(np.sum(arr,axis=1))
print(np.min(arr,axis=0))
print(np.min(arr,axis=1))
```
