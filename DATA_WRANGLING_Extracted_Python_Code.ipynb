```python


import pandas as pd
# 1. HIERARCHICAL INDEXING (MULTIINDEX)


# Student marks by department and semester
data = {
    "Department": ["CSE", "CSE", "CSE", "DS", "DS", "DS"],
    "Semester": [1, 2, 3, 1, 2, 3],
    "Average_Marks": [76, 81, 79, 82, 85, 88]
}

df = pd.DataFrame(data)

multi = df.set_index(["Department", "Semester"])
print("\n--- MultiIndex DataFrame ---")
print(multi)
```

```python
# Select all semester records for CSE
print("\n--- All CSE records ---")
print(multi.loc["CSE"])
```

```python
# Select average marks for CSE, Semester 2
print("\n--- CSE Semester 2 Average Marks ---")
print(multi.loc[("CSE", 2), "Average_Marks"])
```

```python
# Convert index levels back into ordinary columns
print("\n--- Reset Index ---")
print(multi.reset_index())
```

```python
# Other useful MultiIndex operations
print("\n--- Sorted Index ---")
print(multi.sort_index())
```

```python
print("\n--- Swapped Index Levels ---")
print(multi.swaplevel(0, 1))
```

```python
print("\n--- Unstacked Average Marks ---")
print(multi["Average_Marks"].unstack())
```

```python
# 2. COMBINING AND MERGING DATASETS - merge()


students = pd.DataFrame({
    "Student_ID": [101, 102, 103, 104],
    "Name": ["Asha", "Ravi", "Neha", "Imran"],
    "Department": ["CSE", "DS", "CSE", "DS"]
})

marks = pd.DataFrame({
    "Student_ID": [101, 102, 104, 105],
    "Data_Science_Marks": [86, 91, 78, 88]
})

# Keep only IDs present in both datasets (inner join)
data1 = pd.merge(students, marks, on="Student_ID", how="inner")
print("\nInner Merge")
print(data1)

# Keep every student; unmatched marks become NaN (left join)
left = pd.merge(students, marks, on="Student_ID", how="left")
print("\n--- Left Merge ---")
print(left)

# Keep all IDs from both datasets (outer join)
outer = pd.merge(students, marks, on="Student_ID", how="outer")
print("\n--- Outer Merge ---")
print(outer)

# Right join: keep every row from the right table
right = pd.merge(students, marks, on="Student_ID", how="right")
print("Right Merge")
print(right)
```
